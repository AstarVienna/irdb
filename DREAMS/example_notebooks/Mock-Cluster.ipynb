# A mock cluster

With the FWHM of the PSF artificially enlarged so the sources are easy to see.

```python
import os
import pytest
import numpy as np
from astropy.io.fits import HDUList
from matplotlib import pyplot as plt
from matplotlib.colors import LogNorm
import scopesim
import synphot
from scopesim import Source
from astropy.io import fits
from scopesim import rc
from scopesim.source.source_templates import star_field
import scopesim_templates as sim_tp
from scopesim.optics.fov_manager import FOVManager
```

```python
cmds = scopesim.UserCommands(use_instrument="DREAMS")
cmds["!OBS.dit"] = 1000
cmds["!OBS.ndit"] = 1000
cmds["!DET.bin_size"] = 1
cmds["!OBS.sky.bg_mag"] = 14.9
cmds["!OBS.sky.filter_name"] = "J"
cmds["SIM.sub_pixel.flag"] = True
dreams = scopesim.OpticalTrain(cmds)
dreams["detector_linearity"].include = False
```

```python
print("PSF was", dreams["seeing_psf"].meta["fwhm"], "arcseconds")
dreams["seeing_psf"].meta["fwhm"] = 25
print("PSF is", dreams["seeing_psf"].meta["fwhm"], "arcseconds")
```

```python
# Cluster artificially made smaller so it fits in the field of view.
src = sim_tp.stellar.clusters.cluster(mass=10000,  distance=1800, core_radius=5, seed=9002)

dreams.observe(src)
hdus = dreams.readout("LMC.fits")
```

```python
# detector_order = [2, 1, 4, 3, 6, 5]
detector_order = [1, 2, 3, 4, 5, 6]

# Since the sources are now artificially large, they can be seen.
plt.figure(figsize=(20, 20))
for plot_number, hdu_number in enumerate(detector_order, 1):
    plt.subplot(3, 2, plot_number)
    plt.imshow(hdus[0][hdu_number].data, norm='log', cmap="hot")
    plt.colorbar()


plt.show()
```
